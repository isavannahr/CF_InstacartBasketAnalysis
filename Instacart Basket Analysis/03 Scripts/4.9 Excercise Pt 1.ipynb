# 04 Data Consistency Checks

### Mixed Data Types

```python
for col in cust.columns.tolist():
  weird = (cust[[col]].map(type) != cust[[col]].iloc[0].apply(type)).any(axis = 1)
  if len (cust[weird]) > 0:
    print (col)
```

The column first_name appears to have a mixed data type. This column is consist of names onl which are strings so I will make this column string type only.

```python
cust['first_name'] = cust['first_name'].astype('str')
cust['first_name'].dtype
```

### Save Memory

```python
cust.info()
```

```python
cust['user_id']= cust['user_id'].astype('int32')
cust['age']= cust['age'].astype('int8')
cust['n_dependents']= cust['n_dependents'].astype('int8')
cust['income']= cust['income'].astype('int32')
cust['gender'] = cust['gender'].astype('category')
cust['fam_status'] = cust['fam_status'].astype('category')
```

```python
cust.info()
```

```python
cust.head()
```

Looks like values were not affected by truncating data types to be able to merge without data memory errors. 

```python
cust.describe()
```

```python
df.info()
```

```python
df.head()
```

Many of these can be category variables.

```python
df['loyalty_flag']= df['loyalty_flag'].astype('category')
df['price_range_loc']= df['price_range_loc'].astype('category')
df['busiest_day']= df['busiest_day'].astype('category')
df['busiest_days']= df['busiest_days'].astype('category')
df['busiest_period_of_day']= df['busiest_period_of_day'].astype('category')
df['spending_flag']= df['spending_flag'].astype('category')
df['order_frequency_flag']= df['order_frequency_flag'].astype('category')
```

```python
df.sort_values('user_id', inplace=True)
cust.sort_values('user_id', inplace=True)
```

Sorting can also help prevent Memory Errors when merging.

```python
df.info()
```

### Merge

```python
df_merged = df.merge(cust, on = 'user_id', indicator = '_custMerge')
```
