## Deriv03 Deriving New Variables

### Price Label

```python
def price_label(row):

  if row['prices'] <= 5:
    return 'Low-range product'
  elif (row['prices'] > 5) and (row['prices'] <= 15):
    return 'Mid-range product'
  elif row['prices'] > 15:
    return 'High range'
  else: return 'Not enough data'
```

```python
df_subset['price_range'] = df_subset.apply(price_label, axis=1)
```

```python
df_subset['price_range'].value_counts(dropna = False)
```

```python
highend = df_subset.loc[df_subset['prices'] > 15]
highend.shape
```

### Busiest Period of Day

When too many users make Instacart orders at the same time, the app freezes. The senior technical officer at Instacart wants you to identify the busiest hours of the day

```python
df['order_time_of_day'].value_counts(dropna = False)
```

The busiest time of the day is 10 and the least busies time of day is 3.

```python
df.loc[df['order_time_of_day'] == 10, 'busiest_period_of_day'] = 'Most orders'
```

```python
df.loc[(df['order_time_of_day'] != 10) & (df['order_time_of_day'] != 3), 'busiest_period_of_day'] = 'Average orders' 
```

```python
df.loc[df['order_time_of_day'] == 3, 'busiest_period_of_day'] = 'Fewest orders'
```

```python
df['busiest_period_of_day'].value_counts(dropna = False)
```

We expect fewest orders and most orders to match the frequency of 3 and 10 in the order_time_of_day frequency counts.
